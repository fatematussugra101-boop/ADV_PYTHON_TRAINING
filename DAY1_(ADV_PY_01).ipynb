```python
#inheritance
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

obj=B()
obj.showA()
obj.showB()
```

```python
#multi-level inheritance
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C(B):
  def showC(self):
    print("showC function")

obj=C()
obj.showA()
obj.showB()
obj.showC()
```

```python
#multiple inheritance
class A:
  def showA(self):
    print("showA function")

class B():
  def showB(self):
    print("showB function")

class C(B,A):
  def showC(self):
    print("showC function")

obj=C()
obj.showC()
obj.showB()
obj.showA()

```

```python
#hierarchal inheritance
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C(A):
  def showC(self):
    print("showC function")

obj=C()
obj.showC()
obj.showA()
```

```python
#hybrid inheritance
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C():
  def showC(self):
    print("showC function")

class D(B,C):
  def showD(self):
    print("showD function")

obj=D()
obj.showC()
obj.showA()
obj.showD()
obj.showB()

```

```python
class A:
  def showA(self):
    print("showA function")

class B(A):
  def showB(self):
    print("showB function")

class C():
  def showC(self):
    print("showC function")

class D(B,C):
  def showD(self):
    print("showD function")

obj=D()
obj.showC()
obj.showA()
obj.showD()
obj.showB()
```
