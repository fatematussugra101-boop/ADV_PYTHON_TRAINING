```python
#writing data to csv file

import csv
with open("emp.csv","w",newline='') as f:
  w=csv.writer(f)
  w.writerow(["ENO","ENAME","ESALARY","EADDR"])
  n=int(input("enter number of employees"))
  for i in range(n):
    eno=input("enter employee no.:")
    ename=input("enter employee name:")
    esal=input("enter employee salary:")
    eaddr=input("enter employee address:")
    w.writerow([eno,ename,esal,eaddr])

print("data inserted successfully")
```

```python
import csv
f=open("emp.csv",'r')
r=csv.reader(f)
data=list(r)
print(data)  #print in list format
for line in data:
  for word in line:
    print(word,"\t",end='')
  print()
```

NUMPY

```python
import numpy as np
students=np.array([
    [85,90,70],
    [40,78,49],
    [90,70,86],
    [67,99,84],
])
students
```

```python
avg_marks=np.mean(students,axis=1)
print("average marks:",avg_marks)
```

```python
avg_mark=np.mean(students)
print("average marks:",avg_mark)
```

```python
avg=np.mean(avg_marks)
print(avg)
```

```python
topper=np.argmax(avg_marks)
print("topper marks:",topper)
```

```python
topper=np.max(avg_marks)
print("topper marks:",topper)
```

```python
med=np.median(students,axis=1)
print("median marks:",med)
```

```python
failed_student=np.where(avg_marks<50)
print("failed student:",failed_student[0]+1)
```

```python
#performance report
for i, avg in enumerate(avg_marks):
  if avg>50:
    status="pass"
  else:
    status="fail"
  print(f"student{i+1}:Avg={avg:.2f} | {status}")
```
