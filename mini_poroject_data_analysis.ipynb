DATA ANALYSIS

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data={
    'name':['a','b','c','d','e'],
    'math':[78,67,90,56,88],
    'science':[85,72,95,60,84],
    'english':[88,70,92,58,86]
}
df=pd.DataFrame(data)
print(df)
```

```python
marks=df[["math","science","english"]].values
average_marks=np.mean(marks,axis=1)
df["average"]=average_marks
print(df)
```

```python
def grade(avg):
  if avg>=90:
    return 'A'
  elif avg>=75:
    return 'B'
  elif avg>=60:
    return 'C'
  else:
    return 'D'
df['Grade']=df["average"].apply(grade)
df
```

```python
plt.figure()
plt.bar(df["name"],df["average"])
plt.xlabel('Students')
plt.ylabel('Average Marks')
plt.title('Average Marks of Students')
plt.show()
```

```python
# Average marks per subject

df[["math", "science", "english"]].mean().plot(kind="bar")

plt.title("Average Marks per Subject")
plt.ylabel("Marks")
plt.show()
```
